# **Interactive plots**

This scatter plot visualizes the Iris dataset, showing the relationship between sepal width and sepal length, colored by species. It also includes marginal violin and box plots for individual variable distributions and a trendline for linear regression.

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species', marginal_y='violin',
                 marginal_x='box', trendline='ols', template='simple_white')

fig.show(renderer='iframe')
```

This Choropleth map uses the Gapminder dataset for the year 2007 to display population (`pop`) by country (`locations="iso_alpha"`). Hover over countries to see their names and population. The `color_continuous_scale` option sets the color scheme, and `projection="natural earth"` specifies the map projection.

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007")
fig = px.choropleth(df, locations="iso_alpha", color="pop", hover_name="country",
                    color_continuous_scale=px.colors.sequential.Plasma,
                    projection="natural earth")
fig.show(renderer='iframe')
```

This animated scatter plot from the Gapminder dataset visualizes the evolution of GDP per capita (`gdpPercap`) and life expectancy (`lifeExp`) over different years (`animation_frame='year'`). The size of the markers represents population (`pop`), and colors distinguish continents (`color='continent'`). The `facet_col='continent'` option creates separate subplots for each continent, and `log_x=True` applies a logarithmic scale to the x-axis for better visualization of GDP ranges.

```python
import plotly.express as px
df = px.data.gapminder()
fig = px.scatter(df, x='gdpPercap', y='lifeExp', animation_frame='year', animation_group='country',
                 size='pop', color='continent', hover_name='country', facet_col='continent',
                 log_x=True, size_max=45, range_x=[100,100000], range_y=[25,90])
fig.show(renderer='iframe')
```

This pie chart displays the population distribution of European countries in 2007 from the Gapminder dataset. Smaller countries (population less than 2 million) are grouped into 'Other countries' for better readability.

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007").query("continent == 'Europe'")
df.loc[df['pop'] < 2.e6, 'country'] = 'Other countries' #represent only large countries
fig = px.pie(df, values='pop', names='country', title='Population of European continent')
fig.show(renderer='iframe')
```

This sunburst chart visualizes the hierarchical structure of continents and countries, showing the population (`values='pop'`) for each, with colors indicating life expectancy (`color='lifeExp'`). Hovering over a segment reveals additional `iso_alpha` data.

```python
import plotly.express as px
df = px.data.gapminder().query("year == 2007")
fig = px.sunburst(df, path=['continent','country'], values = 'pop',
                  color='lifeExp', hover_data=['iso_alpha'])
fig.show(renderer='iframe')
```
