```python
import pandas as pdimport numpy as npimport matplotlib.pyplot as pltimport seaborn as snsimport plotly.express as pximport plotly.graph_objects as gosns.set_style("whitegrid")
```

<div class='section'>Load Dataset</div>

```python
df = pd.read_csv("chocolate_sales_2025_dataset.csv")df.head()
```

<div class='section'>EDA Visualizations</div>

```python
sns.histplot(df['Revenue_USD'],bins=40,color='#5d4037',kde=True)plt.title("Revenue Distribution")plt.show()
```

```python
sns.boxplot(x=df['Revenue_USD'],color='#6d4c41')plt.show()
```

```python
brand=df.groupby("Brand")["Revenue_USD"].sum().sort_values(ascending=False)sns.barplot(x=brand.index,y=brand.values,palette="copper")plt.xticks(rotation=45)plt.show()
```

```python
ptype=df.groupby("Product_Type")["Revenue_USD"].sum()ptype.plot(kind="bar",color="#4e342e")plt.show()
```

```python
country=df.groupby("Country")["Revenue_USD"].sum().sort_values(ascending=False)country.plot(kind="bar",color="#3e2723")plt.show()
```

```python
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,cmap="YlOrBr")plt.show()
```

<div class='section'>Interactive Plotly Charts</div>

```python
fig=px.histogram(df,x="Revenue_USD",color_discrete_sequence=["#6d4c41"])fig.show()
```

```python
fig=px.bar(df.groupby("Brand",as_index=False)["Revenue_USD"].sum(),x="Brand",y="Revenue_USD",color_discrete_sequence=["#5d4037"])fig.show()
```

```python
fig=px.pie(df,names="Product_Type",values="Revenue_USD",color_discrete_sequence=px.colors.sequential.Brwnyl)fig.show()
```

```python
fig=px.scatter(df,x="Units_Sold",y="Revenue_USD",color_discrete_sequence=["#3e2723"])fig.show()
```

```python
fig=px.sunburst(df,path=["Brand","Product_Type"],values="Revenue_USD",color="Revenue_USD",color_continuous_scale="Brwnyl")fig.show()
```

```python
fig=px.line(df,x="Date",y="Revenue_USD",color_discrete_sequence=["#4e342e"])fig.show()
```

```python
fig=px.choropleth(df,locations="Country",locationmode="country names",color="Revenue_USD",color_continuous_scale="Brwnyl")fig.show()
```
