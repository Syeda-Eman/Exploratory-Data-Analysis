
# 🌍 Global AQI Dataset Analysis (2025)

This notebook performs Exploratory Data Analysis (EDA) on the **Global AQI Dataset 2025**.

## Objectives:
- Load and inspect the dataset
- Perform data cleaning checks
- Generate summary statistics
- Visualize AQI distribution
- Analyze regional pollution trends


```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python

df = pd.read_csv('/kaggle/input/global-aqi/global_aqi_all_countries_2025.csv')
df.head()
```

```python

df.info()
```

```python

df.describe()
```

```python

df.isnull().sum()
```

## 📊 AQI Distribution

```python

plt.figure()
plt.hist(df['AQI'], bins=20)
plt.title("Distribution of AQI Values")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.show()
```

## 🌫 Top 10 Most Polluted Countries (Highest AQI)

```python

top_polluted = df.sort_values(by='AQI', ascending=False).head(10)
top_polluted[['Country', 'AQI']]
```

## 🌎 Average AQI by Region

```python

region_avg = df.groupby('Region')['AQI'].mean().sort_values(ascending=False)
region_avg
```

```python

plt.figure()
region_avg.plot(kind='bar')
plt.title("Average AQI by Region")
plt.xlabel("Region")
plt.ylabel("Average AQI")
plt.show()
```

## 🔥 Correlation Matrix

```python

correlation = df.corr(numeric_only=True)

plt.figure()
plt.imshow(correlation)
plt.colorbar()
plt.title("Correlation Matrix")
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.show()
```
