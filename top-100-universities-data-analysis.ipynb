<div class='section'><h2>⚙️ Import Libraries</h2></div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print('Libraries loaded successfully')
```

<div class='section'><h2>📂 Load Dataset</h2></div>

```python
# Update the path if needed
df = pd.read_csv('/kaggle/input/datasets/syedaeman2212/top-100-universities/top_100_universities_dataset.csv')

df.head()
```

<div class='section'>

```python
df.info()
```

<div class='section'><h2>📈 Statistical Summary</h2></div>

```python
df.describe()
```

<div class='section'><h2>👩‍🎓 Student Population Distribution</h2></div>

```python
plt.hist(df['Total_Students'], bins=20)
plt.title('Distribution of Student Population')
plt.xlabel('Total Students')
plt.ylabel('Frequency')
plt.show()
```

<div class='section'><h2>🏫 Programs vs Faculty Analysis</h2></div>

```python
plt.scatter(df['Programs_Offered'], df['Total_Faculty'])
plt.title('Programs Offered vs Faculty Size')
plt.xlabel('Programs Offered')
plt.ylabel('Total Faculty')
plt.show()
```

<div class='section'><h2>🏛 University Type Distribution</h2></div>

```python
df['University_Type'].value_counts().plot(kind='bar')
plt.title('Public vs Private Universities')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```
