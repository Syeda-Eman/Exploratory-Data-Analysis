```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
df = pd.read_csv("/kaggle/input/mango-prices/mango_prices_dataset.csv")  
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(df["AveragePrice"], bins=30, kde=True)
plt.title("Distribution of Average Mango Prices")
plt.xlabel("Average Price")
plt.show()
```

```python
plt.figure(figsize=(10,4))
df.groupby("Date")["AveragePrice"].mean().plot()
plt.title("Average Mango Price Over Time")
plt.xlabel("Date")
plt.ylabel("Average Price")
plt.show()
```

```python
plt.figure(figsize=(10,4))
df.groupby("Date")["Total Volume"].sum().plot()
plt.title("Total Mango Sales Volume Over Time")
plt.xlabel("Date")
plt.ylabel("Total Volume")
plt.show()
```

```python
mango_types = ["Mango_4046", "Mango_4225", "Mango_4770"]

df[mango_types].mean().plot(kind="bar", figsize=(6,4))
plt.title("Average Sales by Mango Type")
plt.ylabel("Units Sold")
plt.show()
```

```python
bag_types = ["Small Bags", "Large Bags", "Total Bags"]

df[bag_types].mean().plot(kind="bar", figsize=(6,4))
plt.title("Average Mango Sales by Bag Size")
plt.ylabel("Bags Sold")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(
    df[["AveragePrice", "Total Volume", "Mango_4046", "Mango_4225", "Mango_4770"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Between Price and Sales")
plt.show()
```

```python
df["Month"] = df["Date"].dt.month

plt.figure(figsize=(8,4))
sns.boxplot(x="Month", y="AveragePrice", data=df)
plt.title("Monthly Seasonality of Mango Prices")
plt.show()
```

# From the exploratory data analysis of the mango price dataset, several key insights emerge:

## 🥭 Price Behavior

The average mango price remains relatively stable over time, fluctuating within a narrow range.

There are short-term spikes and dips, but no extreme long-term upward or downward trend.

The price distribution is well spread and nearly uniform, indicating controlled market pricing rather than heavy volatility.

## 📈 Sales Volume Trends

Total mango sales volume stays consistently high across the observed period.

Occasional dips appear, likely reflecting seasonal availability or supply chain variations, but overall demand remains strong.

High volume alongside stable prices suggests a well-balanced supply–demand market.

## 📦 Mango Type Analysis (PLU Codes)

Medium-sized mangoes (Mango_4225) record the highest average sales, making them the most popular category.

Small (4046) and large (4770) mangoes show slightly lower but still substantial demand.

Consumer preference appears to favor medium-sized mangoes.

## 🛍️ Packaging Insights

Large bags dominate sales, indicating bulk purchasing behavior.

Small bags contribute significantly less to total sales.

This suggests mangoes are primarily purchased for household stocking or commercial use rather than single-unit consumption.

## 🔗 Price vs Demand Relationship

Correlation analysis shows very weak correlation between price and sales volume.

Changes in price do not strongly affect demand, indicating mangoes behave like a price-inelastic commodity.

## ✅ Overall Insight

The mango market demonstrates stable pricing, strong demand, and predictable purchasing patterns.
These characteristics make the dataset highly suitable for:

📊 Time-series forecasting

🤖 Machine learning price prediction

📈 Demand modeling

🧠 Seasonal trend analysis

This EDA provides a solid foundation for building robust forecasting models such as XGBoost or LSTM in the next stage.
