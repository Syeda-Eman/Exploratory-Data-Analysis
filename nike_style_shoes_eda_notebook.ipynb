<div class='card'><h2>⚙️ Import Libraries</h2></div>

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Libraries loaded')
```

<div class='card'><h2>📂 Load Dataset</h2></div>

```python
# Update path if needed
df = pd.read_csv('/kaggle/input/shoes-sales/shoes_sales_dataset.csv')
df.head()
```

<div class='card'><h2>📊 Dataset Overview</h2></div>

```python
df.info()
```

<div class='card'><h2>📈 Summary Statistics</h2></div>

```python
df.describe()
```

<div class='card'><h2>💰 Revenue by Brand</h2></div>

```python
df.groupby('Brand')['Revenue_USD'].sum().plot(kind='bar')
plt.title('Revenue by Brand')
plt.ylabel('Revenue')
plt.show()
```

<div class='card'><h2>📦 Units Sold Distribution</h2></div>

```python
plt.hist(df['Units_Sold'], bins=20)
plt.title('Units Sold Distribution')
plt.xlabel('Units Sold')
plt.ylabel('Frequency')
plt.show()
```

<div class='card'><h2>📊 Price vs Units Sold</h2></div>

```python
plt.scatter(df['Price_USD'], df['Units_Sold'])
plt.xlabel('Price')
plt.ylabel('Units Sold')
plt.title('Price vs Units Sold')
plt.show()
```
